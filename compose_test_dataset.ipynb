```python
import pandas as pd
```

```python
train_cross_df = pd.read_csv('train_dataset_passed_cross_mining_method_validated_df.csv')
train_cross_df.head()
```

```python
qa_only_train_cross_df = train_cross_df[['question', 'positive', 'hard_negative', 'soft_negative']]
qa_only_train_cross_df.head()
```

```python
len(qa_only_train_cross_df)
```

```python
train_oracle_df = pd.read_csv('train_dataset_passed_oracle_method_vlidated_df.csv')
train_oracle_df.head()
```

```python
qa_only_train_oracle_df = train_oracle_df[['question', 'positive', 'hard_negative', 'soft_negative']]
qa_only_train_oracle_df.head()
```

```python
len(qa_only_train_oracle_df)
```

```python
all_df = pd.concat(
    [
        pd.read_csv("attackqa_parquet_processed_all.csv")[["question", "answer"]],
        pd.read_csv("cybersecurityQAAdataset_only_reviewed_by_expert.csv")[["question", "answer"]],
        pd.read_csv("cybersecurity_qa_updated_column_names.csv")[["question", "answer"]]
    ],
    ignore_index=True
).drop_duplicates().reset_index(drop=True)
all_df.head()
```

```python
len(all_df)
```

```python
all_df.rename(columns={"answer": "positive"}, inplace=True)
all_df.head()
```

```python
all_df["positive"].duplicated().sum()
```

```python
len(pd.concat([qa_only_train_cross_df[['question', 'positive']], qa_only_train_oracle_df[['question', 'positive']]]))
```

```python
exclude_df = pd.concat([
    qa_only_train_cross_df[['question', 'positive']], qa_only_train_oracle_df[['question', 'positive']]
]).drop_duplicates()
exclude_df.head()
```

```python
len(exclude_df)
```

```python
merged_df = pd.merge(all_df, exclude_df, on=['question', 'positive'], how='left', indicator=True)
```

```python
merged_df.head()
```

```python
semi_fineal_test_df = merged_df[merged_df['_merge'] == 'left_only'].drop(columns=['_merge'])
```

```python
semi_fineal_test_df.head()
```

```python
len(semi_fineal_test_df)
```

```python
negatives_cross = pd.concat([
    qa_only_train_cross_df['hard_negative'],
    qa_only_train_cross_df['soft_negative']
]).dropna()
```

```python
negatives_oracle = pd.concat([
    qa_only_train_oracle_df['hard_negative'],
    qa_only_train_oracle_df['soft_negative']
]).dropna()
```

```python
all_negatives = set(pd.concat([negatives_cross, negatives_oracle]))
```

```python
final_test_df = semi_fineal_test_df[
    ~semi_fineal_test_df['positive'].isin(all_negatives)
].reset_index(drop=True)
```

```python
len(final_test_df)
```
