```python
import pandas as pd
```

# attackqa.parquet

```python
attackqa_parquet = pd.read_parquet(
    "https://huggingface.co/datasets/sambanovasystems/attackqa/resolve/main/attackqa.parquet",
    engine='pyarrow'
)
```

```python
attackqa_parquet.head()
```

```python
with pd.option_context('display.max_colwidth', None):
    display(attackqa_parquet[["question", "answer"]])
```

```python
attackqa_human = attackqa_parquet[attackqa_parquet["human_answer"] == True].copy()
attackqa_human = attackqa_human.reset_index(drop=True)
attackqa_human = attackqa_human.dropna(subset=["question", "answer"])
print(f"Всего строк: {len(attackqa_human)}")
print(f"Строк с NaN в question: {attackqa_human['question'].isna().sum()}")
print(f"Строк с NaN в answer: {attackqa_human['answer'].isna().sum()}")
print(f"Строк с NaN в question ИЛИ answer: {attackqa_human[['question', 'answer']].isna().any(axis=1).sum()}")
```

```python
attackqa_parquet.dropna(subset=["question", "answer"]).info()
```

```python
attackqa_parquet.info()
```

```python
attackqa_parquet.describe()
```

```python
attackqa_parquet_human_answered = attackqa_parquet[attackqa_parquet['human_answer'] == True]
attackqa_parquet_human_answered.head()
```

```python
attackqa_parquet_human_answered.info()
```

### Length of questions

```python
(attackqa_parquet_human_answered['question'].str.len().min(),
attackqa_parquet_human_answered['question'].str.len().mean().round(1),
attackqa_parquet_human_answered['question'].str.len().max())
```

### Length of answers

```python
(attackqa_parquet_human_answered['answer'].str.len().min(),
attackqa_parquet_human_answered['answer'].str.len().mean().round(1),
attackqa_parquet_human_answered['answer'].str.len().max())
```

### Words in questions

```python
(attackqa_parquet_human_answered['question'].str.split().apply(len).min(),
attackqa_parquet_human_answered['question'].str.split().apply(len).mean().round(1),
attackqa_parquet_human_answered['question'].str.split().apply(len).max())
```

### Words in answers

```python
(attackqa_parquet_human_answered['answer'].str.split().apply(len).min(),
attackqa_parquet_human_answered['answer'].str.split().apply(len).mean().round(1),
attackqa_parquet_human_answered['answer'].str.split().apply(len).max())
```

# cybersecurityQAAdataset.json

```python
cybersecurityQAAdataset_json = pd.read_json("https://huggingface.co/datasets/Rowden/CybersecurityQAA/resolve/main/cybersecurityQAAdataset.json")
```

```python
cybersecurityQAAdataset_json.head()
```

```python
cybersecurityQAAdataset = pd.json_normalize(cybersecurityQAAdataset_json['vars'])
cybersecurityQAAdataset.head()
```

```python
with pd.option_context('display.max_colwidth', None):
    display(cybersecurityQAAdataset[["question", "answer"]].head())
```

```python
cybersecurityQAAdataset.info()
```

```python
cybersecurityQAAdataset.describe()
```

```python
cybersecurityQAAdataset['expert_score'] = pd.to_numeric(cybersecurityQAAdataset['expert_score'])
cybersecurityQAAdataset['reviewed_by_expert'] = cybersecurityQAAdataset['reviewed_by_expert'].astype(str).map({'TRUE': True, 'FALSE': False, '': False}).fillna(False)

cybersecurityQAAdataset['reviewed_by_expert'].unique()
```

```python
cybersecurityQAAdataset_only_reviewed_by_expert = cybersecurityQAAdataset[cybersecurityQAAdataset['reviewed_by_expert'] == True]
```

### Length of questions

```python
(cybersecurityQAAdataset_only_reviewed_by_expert['question'].str.len().min(),
cybersecurityQAAdataset_only_reviewed_by_expert['question'].str.len().mean().round(1),
cybersecurityQAAdataset_only_reviewed_by_expert['question'].str.len().max())
```

### Length of answers

```python
(cybersecurityQAAdataset_only_reviewed_by_expert['answer'].str.len().min(),
cybersecurityQAAdataset_only_reviewed_by_expert['answer'].str.len().mean().round(1),
cybersecurityQAAdataset_only_reviewed_by_expert['answer'].str.len().max())
```

```python
cybersecurityQAAdataset_only_reviewed_by_expert['expert_score'].unique()
```

### Words in questions

```python
(cybersecurityQAAdataset_only_reviewed_by_expert['question'].str.split().apply(len).min(),
cybersecurityQAAdataset_only_reviewed_by_expert['question'].str.split().apply(len).mean().round(1),
cybersecurityQAAdataset_only_reviewed_by_expert['question'].str.split().apply(len).max())
```

### Words in answers

```python
(cybersecurityQAAdataset_only_reviewed_by_expert['answer'].str.split().apply(len).min(),
cybersecurityQAAdataset_only_reviewed_by_expert['answer'].str.split().apply(len).mean().round(1),
cybersecurityQAAdataset_only_reviewed_by_expert['answer'].str.split().apply(len).max())
```
